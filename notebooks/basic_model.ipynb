### Market Maker Class
=============================================================================
This market maker will perform the following functions
1. Receive market history to determine the latest bid - offer prices
2. The market maker will then present a quote which is exactly the same as the bid-offer prices
3. I will implement inventory and profit & loss tracking for this market maker

The market maker will be setup to have configurable limits for profit &  loss as well as inventory to limit the magnitude of losses allowed. 

```python
class PrimitiveMarketMaker:
    def __init__(self, best_bid, best_offer, order_size, max_inventory):
        self.best_bid = best_bid
        self.best_offer = best_offer
        self.order_size = order_size
        self.max_inventor = max_inventory

        #State tracking
        self.inventory = 0
        self.cash = 0
        self.history = []

        def quote(self):
            return self.best_bid, self.best_offer
```
